```python
import os
import csv
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
os.makedirs("figures", exist_ok=True)
os.makedirs("results", exist_ok=True)

# Water at 25 C
RHO0 = 997.0          # density, kg/m3
C0 = 1497.0           # speed of sound, m/s
ETA = 0.890e-3        # viscosity, Pa s
KAPPA0 = 1.0 / (RHO0 * C0**2)
KB, TEMP = 1.380649e-23, 298.15
KT = KB * TEMP

# Particle materials (density kg/m3, compressibility 1/Pa)
# EV values are an assumption (lipid vesicle, density ~1.10 g/mL)
MATERIALS = {
    "Polystyrene bead": dict(rho=1050.0, kappa=2.49e-10, color="#1f77b4"),
    "Cell": dict(rho=1070.0, kappa=4.00e-10, color="#2ca02c"),
    "EV (assumed)": dict(rho=1100.0, kappa=4.00e-10, color="#d62728"),
}

def contrast_factor(rho, kappa):
    f1 = 1.0 - kappa / KAPPA0
    f2 = 2.0 * (rho - RHO0) / (2.0 * rho + RHO0)
    return f1 / 3.0 + f2 / 2.0

for name, m in MATERIALS.items():
    m["phi"] = contrast_factor(m["rho"], m["kappa"])
    print(f"{name:20s} contrast factor = {m['phi']:.3f}")

# Channel: width sets the half wave resonance frequency
W = 375e-6
H = 150e-6
F0 = C0 / (2 * W)
print(f"\nChannel {W*1e6:.0f} x {H*1e6:.0f} um, resonance {F0/1e6:.2f} MHz")
```

```python
def k_of(f):
    return 2 * np.pi * f / C0

def boundary_layer(f):
    return np.sqrt(2 * ETA / (RHO0 * 2 * np.pi * f))

def rad_velocity_amp(a, phi, f, E):
    """Speed caused by the sound push (radiation force / drag)."""
    return (2.0 / 3.0) * phi * a**2 * k_of(f) * E / ETA

def streaming_amp(f, E):
    """Speed of the swirling flow the sound creates (Rayleigh streaming)."""
    return 0.75 * E / (RHO0 * C0)

def critical_radius(f, phi):
    """Size where the sound push equals the streaming flow."""
    return 0.75 * boundary_layer(f) / np.sqrt(phi)

def velocity(y, z, a, phi, f, E, h=H):
    k = k_of(f)
    zeta = 2 * z / h
    U = streaming_amp(f, E)
    uy_st = U * np.sin(2 * k * y) * (1 - 3 * zeta**2)
    uz_st = -U * (k * h / 2) * np.cos(2 * k * y) * (zeta - zeta**3)
    uy_rad = rad_velocity_amp(a, phi, f, E) * np.sin(2 * k * y)
    return uy_st + uy_rad, uz_st

def simulate(diams, phi, E, f=F0, n=300, t_end=30.0, dt=2e-3, keep_paths=False):
    """Move n particles per size step by step: forces + random Brownian jiggle."""
    a = np.repeat(np.asarray(diams) / 2, n)
    N = a.size
    y = rng.uniform(a, W - a)
    z = rng.uniform(-H / 2 + a, H / 2 - a)
    D = KT / (6 * np.pi * ETA * a)
    s = np.sqrt(2 * D * dt)
    steps = int(t_end / dt)
    paths = []
    rec_every = max(1, steps // 400)
    for i in range(steps):
        vy, vz = velocity(y, z, a, phi, f, E)
        y = y + vy * dt + s * rng.standard_normal(N)
        z = z + vz * dt + s * rng.standard_normal(N)
        lo, hi = a, W - a
        y = np.where(y < lo, 2 * lo - y, y); y = np.where(y > hi, 2 * hi - y, y)
        zl, zh = -H / 2 + a, H / 2 - a
        z = np.where(z < zl, 2 * zl - z, z); z = np.where(z > zh, 2 * zh - z, z)
        if keep_paths and i % rec_every == 0:
            paths.append(np.stack([y, z]))
    return a, y, z, (np.array(paths) if keep_paths else None)

def focused_fraction(y, band=0.1):
    """Share of particles that ended up in the centre of the channel."""
    return np.mean(np.abs(y - W / 2) < band * W)

print("Physics functions ready")
```

```python
diams = np.logspace(np.log10(50e-9), np.log10(10e-6), 13)
E_RUN = 10.0              # acoustic energy density, J/m3
energies = [10.0, 100.0]
phi_ev = MATERIALS["EV (assumed)"]["phi"]
rows, frac = [], {}

for mat in MATERIALS:
    phi = MATERIALS[mat]["phi"]
    a, y, z, _ = simulate(diams, phi, E_RUN)
    fr = [focused_fraction(y[a == d / 2]) for d in diams]
    frac[(mat, E_RUN)] = np.array(fr)
    for d, v in zip(diams, fr):
        rows.append(dict(material=mat, E_ac=E_RUN, diameter_um=d * 1e6, focused_fraction=v))
    print(f"{mat:18s}: " + " ".join(f"{d*1e6:.2g}um:{v:.2f}" for d, v in zip(diams, fr)))

with open("results/focusing_vs_size.csv", "w", newline="") as fh:
    wr = csv.DictWriter(fh, fieldnames=rows[0].keys()); wr.writeheader(); wr.writerows(rows)

def crossover_diameter(fr, d, level=0.6):
    for i in range(1, len(d)):
        if fr[i - 1] < level <= fr[i]:
            t = (level - fr[i - 1]) / (fr[i] - fr[i - 1])
            return 10 ** (np.log10(d[i - 1]) + t * (np.log10(d[i]) - np.log10(d[i - 1])))
    return np.nan

print("\nCritical diameter, simulation vs formula:")
for mat in MATERIALS:
    dc = crossover_diameter(frac[(mat, E_RUN)], diams)
    print(f"  {mat:18s} sim {dc*1e6:.2f} um, formula {2*critical_radius(F0, MATERIALS[mat]['phi'])*1e6:.2f} um")
```

```python
traj = {}
for d in [0.1e-6, 1e-6, 5e-6]:
    _, _, _, p = simulate([d], phi_ev, 10.0, n=12, t_end=60.0, keep_paths=True)
    traj[d] = p
print("Trajectories done")
```

```python
freqs = np.logspace(np.log10(0.5e6), np.log10(2e9), 200)
for name, m in MATERIALS.items():
    m["dc_curve"] = 2 * critical_radius(freqs, m["phi"])
    fn = F0 * (2 * critical_radius(F0, m["phi"]) / 100e-9) ** 2
    print(f"{name:18s} frequency needed for 100 nm: {fn/1e6:.0f} MHz")

def focus_time(d, phi, E, f=F0, y0=W / 8, y_end=W / 2 - 5e-6):
    k = k_of(f)
    tstar = 3 * ETA / (4 * phi * (k * d / 2) ** 2 * E)
    return tstar * np.log(np.tan(k * y_end) / np.tan(k * y0))

Es = np.logspace(0, 3, 100)
ft = {
    "10 um cell": focus_time(10e-6, MATERIALS["Cell"]["phi"], Es),
    "1 um particle": focus_time(1e-6, MATERIALS["Cell"]["phi"], Es),
    "100 nm EV": focus_time(100e-9, phi_ev, Es),
}
print("\nFocusing time at 10 J/m3:")
for kname, v in ft.items():
    print(f"  {kname:14s} {np.interp(10, Es, v):10.1f} s")

for E in energies:
    dU = 4 * np.pi * phi_ev * (50e-9) ** 3 * E
    print(f"100 nm EV trap depth at {E:.0f} J/m3: {dU/KT:.1f} kT")
```

```python
fig, ax = plt.subplots(2, 2, figsize=(11, 8.2))

Y, Z = np.meshgrid(np.linspace(0, W, 40), np.linspace(-H / 2, H / 2, 20))
uy, uz = velocity(Y, Z, 0.0, 0.0, F0, 10.0)
ax[0, 0].streamplot(Y * 1e6, Z * 1e6, uy, uz, color="0.75", density=1.0, linewidth=0.7)
cols = {0.1e-6: "#d62728", 1e-6: "#ff7f0e", 5e-6: "#1f77b4"}
for d, p in traj.items():
    for j in range(p.shape[2]):
        ax[0, 0].plot(p[:, 0, j] * 1e6, p[:, 1, j] * 1e6, color=cols[d], lw=0.8, alpha=0.8)
    ax[0, 0].plot([], [], color=cols[d], label=f"{d*1e6:g} um")
ax[0, 0].axvline(W / 2 * 1e6, color="k", ls=":", lw=0.8)
ax[0, 0].set(xlabel="Channel width (um)", ylabel="Height (um)",
             title=f"(a) Trajectories, 60 s, {F0/1e6:.1f} MHz, 10 J/m$^3$")
ax[0, 0].legend(loc="upper right", fontsize=8)

for (mat, E), fr in frac.items():
    ax[0, 1].semilogx(diams * 1e6, fr, "-", marker="o", ms=3, color=MATERIALS[mat]["color"], label=mat)
ax[0, 1].axvspan(0.03, 0.15, color="#d62728", alpha=0.1)
ax[0, 1].text(0.035, 0.9, "small EVs", color="#d62728", fontsize=8)
ax[0, 1].axhline(0.2, color="0.5", ls=":", lw=0.8)
ax[0, 1].set(xlabel="Particle diameter (um)", ylabel="Fraction in central band after 30 s",
             title="(b) Focusing vs size and material, 10 J/m$^3$", ylim=(0, 1.02))
ax[0, 1].legend(fontsize=7)

for name, m in MATERIALS.items():
    ax[1, 0].loglog(freqs / 1e6, m["dc_curve"] * 1e6, color=m["color"], label=name)
ax[1, 0].axhspan(0.03, 0.15, color="#d62728", alpha=0.1)
ax[1, 0].axvline(F0 / 1e6, color="k", ls=":", lw=0.8)
ax[1, 0].set(xlabel="Frequency (MHz)", ylabel="Critical diameter (um)",
             title="(c) Smallest focusable size vs frequency")
ax[1, 0].legend(fontsize=8)

for (kname, v), c in zip(ft.items(), ["#2ca02c", "#ff7f0e", "#d62728"]):
    ax[1, 1].loglog(Es, v, color=c, label=kname)
ax[1, 1].axhline(60, color="0.5", ls=":", lw=0.8)
ax[1, 1].set(xlabel="Acoustic energy density (J/m$^3$)", ylabel="Focusing time (s)",
             title="(d) Radiation only focusing time vs power")
ax[1, 1].legend(fontsize=8)

fig.tight_layout()
fig.savefig("figures/acoustofluidic_ev_study.png", dpi=200)
plt.show()
```
